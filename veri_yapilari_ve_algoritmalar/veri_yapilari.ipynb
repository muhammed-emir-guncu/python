```python
class BiNode:
    def __init__(self,idd,left=None,right=None):
        self.idd=idd
        self.left=left
        self.right=right
    
    def __repr__(self):
        ll=repr(self.left)+' < ' if self.left else ''
        rr=' > '+repr(self.right) if self.right else ''
        return f"[{ll}{self.idd}{rr}]"
    
    def search(self,a,rot=''):
        if a== self.idd:
            return True,rot
        elif self.right and a>self.idd:
            return self.right.search(a,rot+'r')
        elif self.left and a<self.idd:
            return self.left.search(a,rot+'l')
        else :
            return False,''
        
    def add(self,b):
        if isinstance(b,BiNode):
            a=b.idd
        else :
            a=b
            b=BiNode(b)
        if a >self.idd:
            if self.right:
                self.right.add(b)
            else:
                self.right=b
        elif a < self.idd :
            if self.left:
                self.left.add(b)
            else:
                self.left=b
        
    def travel(self,xs):
        if len(xs) == 0:
            return self
        else:
            for i in xs:
                if i =='l' and self.left:
                    return self.left.travel(xs[1::])
                elif i =='r' and self.right:
                    return self.right.travel(xs[1::])
                else :
                    raise 'IndexError'


```

```python
a=BiNode(7, BiNode(4), BiNode(9, BiNode(8)))
print(a)
a.add(5)
print(a)
print(a.travel('lr').idd)
a.add(BiNode(10))
print(a)
```

```python
class Linked_list:
    def __init__(self,value,node=None):
        self.value:any= value
        self.node:Linked_list=node
    def __repr__(self):
        return f"{self.value} {'-> '+repr(self.node) if self.node else ''}"
    def last(self):
        if self.node:
            return self.node.last()
        else:
            return self
    def __add__(self,other):
        a=self.clone()
        if not isinstance(other,Linked_list):
            a.last().node=Linked_list(other)
        else:
            a.last().node=other
        return a
    def __radd__(self,other):
        return Linked_list(other,self.clone())
    def __iadd__(self,other):
         
        if not isinstance(other,Linked_list):
            self.last().node=Linked_list(other)
        else:
            self.last().node=other
        return self
    def __iter__(self):
        current = self
        yield current.value
        while current:=current.node:
            yield current.value
    def __len__(self):
        l=0
        for _ in self:
            l+=1 
        return l
    def __contains__(self,a):
        for i in self:
            if i ==a:
                return True  
        return False
    def __getitem__(self,s):
        i=0
        for x in self:
            if i ==s :
                return x
            i+=1
        raise 'Index out of range'
    def clone(self):
        nod = self.node.clone() if self.node else None
        return Linked_list(self.value, nod)
    def head_sep(self):
        return self.value, self.node
    def revr(self):
        a=None
        for i in self:
            a=Linked_list(i,a)
        return a
    def last_sep(self):
        a=self.revr()
        a,b=a.head_sep()
        return b.revr(), a

                    
```

```python
def bs(xs,a):
    if len(xs)==1:
        return xs[0]==a
    elif len(xs)==2:
        return xs[0]==a or xs[1]==a
    else:
        l=int(len(xs)/2)
        if a>=xs[l]:
            return bs(xs[l:],a)
        elif a<xs[l]:
            return bs(xs[:l],a)

print(bs([1,3,6,12,18,20],10))
```

```python
def bs(xs, a):
    low, high = 0, len(xs) - 1
    while low <= high:
        mid = (low + high) // 2
        if xs[mid] == a:
            return True
        elif xs[mid] < a:
            low = mid + 1
        else:
            high = mid - 1
    return False
print(bs([1,3,6,12,18,20],10))
```

```python
class Stack:
    class Cont:
        def __init__(self,value,node=None):
            self.value=value
            self.node=node
        def __repr__(self):
            return f"{self.value}{', ' + repr(self.node) if self.node else '' }"
    
    def __init__(self):
        self.cont=None
    def __repr__(self):
        return f'[{self.cont if self.cont else '-'}]'
    def push(self,val):
        self.cont=self.Cont(val,self.cont)
    def pop(self):
        if not self.is_empty():
            a=self.cont.value
            self.cont=self.cont.node
            return a
        else:
            raise IndexError('stack is empty')
    def peek(self):
        if not self.is_empty():
            return self.cont.value
        else:
            raise IndexError('stack is empty')
    def is_empty(self):
        return self.cont == None
```

```python
a=Stack()
print(a, 'liste bos mu', a.is_empty())
a.push(12)
a.push(13)
a.push(14)
b=a.pop()
print(a.is_empty())
print(a.peek())
print(a,b)

```

```python
def inorder(xs:list)->list:
    if not xs:
        return []
    elif len(xs) == 1:
        return xs
    else:
        r=int(len(xs)/2)
        return [xs[r]] + inorder(xs[:r])+ inorder(xs[r+1:]) 
```
